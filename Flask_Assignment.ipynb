1. What is a RESTful API?
- A RESTful API is an application programming interface (API) that follows REST (Representational State Transfer) principles. It uses standard HTTP methods (GET, POST, PUT, DELETE) to allow communication between clients and servers in a stateless manner.

2. Explain the concept of API specification
- An API specification defines how an API behaves, including its endpoints, request/response format, authentication requirements, and expected errors. Standards like OpenAPI help document APIs for developers.

3. What is Flask, and why is it popular for building APIs?
- Flask is a lightweight Python web framework that is widely used for building APIs due to its simplicity, flexibility, and extensive ecosystem of extensions

4. What is routing in Flask?
- Routing in Flask refers to mapping URLs to specific functions so that different parts of the application can be accessed via predefined paths

5. How do you create a simple Flask application?
- from flask import Flask
apA simple Flask application starts by creating an instance of the Flask class. This instance represents the web application. Then, routes are defined using decorators, which associate URLs with specific functions. When a request matches a route, the corresponding function executes and returns a response. The application runs using the app.run() method.

6. What are HTTP methods used in RESTful APIs?
- GET, POST, PUT, DELETE, PATCH. This methods used in RESTful APIs.

7. What is the purpose of the @app.route() decorator in Flask?
- It binds a URL to a Python function, enabling Flask to handle requests at that endpoint.

8. What is the difference between GET and POST HTTP methods?
- GET requests retrieve data from a server. and POST requests send data to the server, typically used for creating new records

9. How do you handle errors in Flask APIs?
- Flask provides built-in error handling mechanisms, allowing developers to define custom error responses for different HTTP status codes. By using the @app.errorhandler() decorator, applications can handle specific errors (such as 404 Not Found) gracefully, returning JSON-formatted messages instead of generic error pages

10. How do you connect Flask to a SQL database?
- To connect Flask to a SQL database, the Flask-SQLAlchemy extension is commonly used. SQLAlchemy is an Object Relational Mapper (ORM) that simplifies interactions with databases. Flask applications specify the database connection URL in the configuration, and SQLAlchemy manages database interactions, enabling developers to use Python code instead of direct SQL queries.

11. What is the role of Flask-SQLAlchemy?
- Flask-SQLAlchemy is an ORM (Object Relational Mapper) that allows Flask applications to interact with databases using Python classes.

12. What are Flask blueprints, and how are they useful?
- Blueprints help organize large applications by grouping related routes into modules

13. What is the purpose of Flask's request object?
- It allows access to incoming request data, including headers, query parameters, and JSON payload.

14. How do you create a RESTful API endpoint using Flask?
- A RESTful API endpoint is created by defining a route in Flask and specifying the allowed HTTP methods. When a request is sent to the defined route, Flask invokes the corresponding function, processes the request, and returns a response, often in JSON format.

15. What is the purpose of Flask's jsonify() function?
- It converts Python dictionaries into JSON responses.

16. Explain Flask’s url_for() function.
- The url_for() function dynamically generates URLs based on function names rather than hardcoding them. This ensures that if the route structure changes, the generated URLs remain valid. It helps maintain flexibility in web applications by creating URLs programmatically

19. What are HTTP status codes, and why are they important in a Flask API?

```python
# 5. How can you generate URLs for routes in Flask using url_for?
from flask import Flask, url_for

app = Flask(__name__)

@app.route('/')
def home(): return "Welcome!"

@app.route('/user/<username>')
def user_profile(username): return f"User: {username}"

@app.route('/generate_url')
def generate_url():
    return f"Home: {url_for('home')} <br> User: {url_for('user_profile', username='JohnDoe')}"

if __name__ == '__main__':
    app.run()
```

```python
# 6. How do you handle forms in Flask?

from flask import Flask, request, render_template

app = Flask(__name__)

@app.route('/form', methods=['GET', 'POST'])
def form():
    if request.method == 'POST':
        username = request.form.get('username', 'Guest')
        return f"Hello, {username}!"
    return render_template('form.html')

if __name__ == '__main__':
    app.run()
```

```python
# 7. How can you validate form data in Flask?

from flask import Flask, render_template, request
from wtforms import Form, StringField, validators

app = Flask(__name__)
app.config['SECRET_KEY'] = 'your_secret_key'

# Define the form class
class MyForm(Form):
    username = StringField('Username', [validators.Length(min=4, max=25)])

@app.route('/form', methods=['GET', 'POST'])
def form():
    form = MyForm(request.form)
    if request.method == 'POST' and form.validate():
        return f"Hello, {form.username.data}!"
    return render_template('form.html', form=form)

if __name__ == '__main__':
    app.run()
```

```python
# 8. How do you manage sessions in Flask?
from flask import Flask, session, redirect, url_for

app = Flask(__name__)
app.secret_key = 'your_secret_key'  # Required for session management

@app.route('/set_session')
def set_session():
    session['user'] = 'JohnDoe'
    return "Session set!"

@app.route('/get_session')
def get_session():
    user = session.get('user', 'Guest')
    return f"Logged in as {user}"

@app.route('/clear_session')
def clear_session():
    session.pop('user', None)
    return "Session cleared!"

if __name__ == '__main__':
    app.run()
```

```python
# 9. How do you redirect to a different route in Flask?
from flask import Flask, redirect, url_for

app = Flask(__name__)

@app.route('/')
def home():
    return "Welcome to Home Page"

@app.route('/redirect_example')
def redirect_example():
    return redirect(url_for('home'))  # Redirects to the 'home' route

if __name__ == '__main__':
    app.run()
```

```python
# 12. How do you define a custom Jinja filter in Flask?
from flask import Flask

app = Flask(__name__)

# Define a custom filter
@app.template_filter('reverse')
def reverse_filter(s):
    return s[::-1]  # Reverses the string

@app.route('/')
def home():
    return f"Reversed: {{ 'Flask' | reverse }}"  # Jinja will use the custom filter

if __name__ == '__main__':
    app.run()
```

```python
# 13. How can you redirect with query parameters in Flask?
from flask import Flask, redirect, url_for, request

app = Flask(__name__)

@app.route('/redirect_with_params')
def redirect_with_params():
    return redirect(url_for('welcome', user='JohnDoe'))

@app.route('/welcome')
def welcome():
    user = request.args.get('user', 'Guest')  # Retrieve query parameter
    return f"Welcome, {user}!"

if __name__ == '__main__':
    app.run()
```

```python
# 14. How do you return JSON responses in Flask?
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/json')
def json_response():
    data = {"message": "Hello, JSON!", "status": "success"}
    return jsonify(data)  # Converts dictionary to JSON response

if __name__ == '__main__':
    app.run()
```
